```python
import pandas as pd
import uuid
```

```python
df = pd.read_parquet('../parquet_output/events_normalized.parquet')
df.head(10)
```

```python
df['event_timestamp'] = pd.to_datetime(df['event_timestamp'], unit='us')
df
```

```python
df = df.sort_values(['user_pseudo_id', 'event_timestamp'])
df
```

```python
session_timeout = pd.Timedelta(minutes=30)
session_timeout
```

```python
df['prev_timestamp'] = df.groupby('user_pseudo_id')['event_timestamp'].shift()
df
```

```python
df['time_diff'] = pd.to_datetime(df['event_timestamp']) - pd.to_datetime(df['prev_timestamp'])
df
```

```python
df['new_session'] = (df['time_diff'] > session_timeout) | (df['time_diff'].isna())
df
```

```python
df['session_id'] = df.groupby('user_pseudo_id')['new_session'].cumsum()
df
```

```python
df['session_id'] = df['new_session'].apply(lambda x: str(uuid.uuid4()) if x else None)
df['session_id'] = df.groupby('user_pseudo_id')['session_id'].ffill()
```

```python
df.drop(columns=['prev_timestamp', 'time_diff', 'new_session'], inplace=True)
```

```python
df
```

```python
df['session_id'].value_counts()
```

```python
cols = list(df.columns)

# remove session_id and reinsert it after event_timestamp
cols.insert(cols.index('event_timestamp') + 1, cols.pop(cols.index('session_id')))

df = df[cols]
```

```python
df
```

```python
df.to_parquet('../parquet_output/events_session_identifier.parquet', index=False)
```
