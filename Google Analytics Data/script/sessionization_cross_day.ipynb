```python
import pandas as pd
import time 
import datetime
```

```python
start_timestamp = time.time()
```

```python
sessionized_df =pd.read_parquet("../parquet_output/sessionized_data.parquet")
sessionized_df.head()  
```

```python
df = sessionized_df
df
```

```python
df['event_timestamp'] = pd.to_datetime(df['event_timestamp'])
df['event_date'] = df['event_timestamp'].dt.date

df
```

```python
df['minutes_to_midnight'] = (
    pd.to_datetime(df['event_date']) + pd.Timedelta(days=1) - df['event_timestamp']
).dt.total_seconds() / 60

df
```

```python
df.sort_values(by=['user_pseudo_id', 'event_date', 'event_timestamp'], inplace=True)

df
```

```python
last_event_mask = ~df.duplicated(subset=['user_pseudo_id', 'event_date'], keep='last')
last_event_mask
```

```python
df['possible_next_day_session'] = False
```

```python
df.loc[last_event_mask & (df['minutes_to_midnight'] <= 30), 'possible_next_day_session'] = True
```

```python
df.drop(columns=['event_date', 'minutes_to_midnight'], inplace=True)
```

```python
df
```

```python
# last_event_indices = (
#     df.groupby(['user_pseudo_id', 'event_date'])['event_timestamp']
#     .idxmax()
# )

# last_event_indices
```

```python
df['time_of_day'] = df['event_timestamp'].dt.time
```

```python
df['event_date'] = df['event_timestamp'].dt.date
df['minutes_from_midnight'] = (
    df['event_timestamp'] - pd.to_datetime(df['event_date'])
).dt.total_seconds() / 60
```

```python
first_event_indices = (
    df.groupby(['user_pseudo_id', 'event_date'])['event_timestamp']
    .idxmin()
)
```

```python
df['possible_previous_day_session'] = False
```

```python
df.loc[first_event_indices, 'possible_previous_day_session'] = (
    df.loc[first_event_indices, 'minutes_from_midnight'] <= 30
)
```

```python
df
```

```python
df.drop(columns=['time_of_day', 'event_date', 'minutes_from_midnight'], inplace=True)
```

```python
df
```

```python
df = df.sort_values(['user_pseudo_id', 'event_timestamp'])
df
```

```python
df['next_event_timestamp'] = df.groupby('user_pseudo_id')['event_timestamp'].shift(-1)
df
```

```python
df['next_event_date'] = pd.to_datetime(df['next_event_timestamp'].dt.date)
df
```

```python
df['time_to_next_event'] = (df['next_event_timestamp'] - df['event_timestamp']).dt.total_seconds() / 60
df
```

```python
df['is_next_day'] = df['next_event_date'] == (pd.to_datetime(df['event_timestamp'].dt.date) + pd.Timedelta(days=1))
df
```

```python
cross_day_conditions = (
    df['possible_next_day_session'] & 
    df['possible_previous_day_session'].shift(-1).fillna(False) & 
     (df['time_to_next_event'] <= 30) & 
    df['is_next_day']
)

cross_day_conditions
```

```python
df['cross_day_session'] = False
df.loc[cross_day_conditions, 'cross_day_session'] = True
df.loc[cross_day_conditions.shift(1).fillna(False), 'cross_day_session'] = True
df
```

```python
df.drop(['next_event_timestamp', 'next_event_date', 'time_to_next_event', 'is_next_day'], 
        axis=1, inplace=True)
```

```python
df
```

```python
df.to_parquet('../parquet_output/sessionization_cross_day.parquet', index=False)
```
