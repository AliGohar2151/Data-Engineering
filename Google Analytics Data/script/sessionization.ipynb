```python
df = df.sort_values(['user_pseudo_id', 'event_timestamp'])
df
```

```python
session_timeout = pd.Timedelta(minutes=30)
session_timeout
```

```python
df['prev_timestamp'] = df.groupby('user_pseudo_id')['event_timestamp'].shift()
df
```

```python
df['time_diff'] = pd.to_datetime(df['event_timestamp']) - pd.to_datetime(df['prev_timestamp'])
df
```

```python
df['new_session'] = (df['time_diff'] > session_timeout) | (df['time_diff'].isna())
df
```

```python
df['session_id'] = df.groupby('user_pseudo_id')['new_session'].cumsum()
df
```

```python
df['session_id'] = df['user_pseudo_id'].astype(str) + '_' + df['session_id'].astype(str)
df
```

```python
df.drop(columns=['prev_timestamp', 'time_diff', 'new_session'], inplace=True)
df
```

```python
def validate_sessions(df, session_timeout_minutes=30):
    df['event_timestamp'] = pd.to_datetime(df['event_timestamp'])
    
    df = df.sort_values(by=['user_pseudo_id', 'event_timestamp']).copy()
    
    errors = []

    for user_id, group in df.groupby('user_pseudo_id'):
        prev_time = None
        prev_session = None
        
        for idx, row in group.iterrows():
            curr_time = row['event_timestamp']
            curr_session = row['session_id']
            
            if prev_time is not None:
                time_diff = curr_time - prev_time
                
                if time_diff > pd.Timedelta(minutes=session_timeout_minutes):
                    if curr_session == prev_session:
                        errors.append({
                            'index': idx,
                            'user': user_id,
                            'prev_time': prev_time,
                            'curr_time': curr_time,
                            'time_diff': time_diff,
                            'session_error': 'Same session_id despite >30min gap'
                        })
                else:
                    if curr_session != prev_session:
                        errors.append({
                            'index': idx,
                            'user': user_id,
                            'prev_time': prev_time,
                            'curr_time': curr_time,
                            'time_diff': time_diff,
                            'session_error': 'Different session_id despite <=30min gap'
                        })

            prev_time = curr_time
            prev_session = curr_session

    if not errors:
        print("✅ All session_id assignments are correct.")
    else:
        print(f"❌ Found {len(errors)} session_id assignment issues.")
        return pd.DataFrame(errors)

error_df = validate_sessions(df)
if error_df is not None:
    error_df.head()
```
