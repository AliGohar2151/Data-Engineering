```python
import pandas as pd
import uuid
```

```python
df = pd.read_parquet('../parquet_output/events_pageview_identifier.parquet')
df.head()
```

```python
keys = (
    df["session_id"].astype(str) + "_" +
    df["pageview_id"].astype(str) + "_" +
    df["event_name"].astype(str) + "_" +
    df["event_timestamp"].astype(str)
)

df["event_id"] = [str(uuid.uuid5(uuid.NAMESPACE_DNS, k)) for k in keys]
```

```python
df
```

```python
cols = list(df.columns)

desired_order =['event_date','event_timestamp','session_id','pageview_id','event_name','event_id']
new_order = desired_order + [c for c in cols if c not in desired_order]
df = df[new_order]
df
```

```python
df.to_parquet('../parquet_output/events_event_identifier.parquet')
```
