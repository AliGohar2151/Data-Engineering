```python
import pandas as pd 
import uuid
```

```python
df = pd.read_parquet('../ga4-clean/session-identifier/analytics291746817/2024/10/events_session_identifier.parquet')
df
```

```python
groupby_df = df.groupby("session_id").agg(
    total_events=("session_id", "count"),                    # count of events in session
    first_event=("event_timestamp", "min"),                  # session start
    last_event=("event_timestamp", "max"),                   # session end
    session_duration=("event_timestamp", lambda x: x.max() - x.min()),
    unique_pages=("ep_page_location", "nunique"),            # how many unique pages
    first_page=("ep_page_location", "first"),                # first page visited
    last_page=("ep_page_location", "last")                   # last page visited
).reset_index()

groupby_df
```

```python
df["is_missing"] = (
    df["ep_page_location"].isna() |
    (df['ep_page_location']=='None')
    (df["ep_page_location"] == "")
)
df
```
